```python
import zipfile

filelist_content = """0_Dataset.ipynb
1_Preliminaries.ipynb
2_Training.ipynb
3_Inference.ipynb
0_Dataset.html
1_Preliminaries.html
2_Training.html
3_Inference.html
model.py
data_loader.py
data_loader_val.py
vocabulary.py
"""
with open('filelist.txt', 'w') as f:
    f.write(filelist_content)
print('filelist.txt updated')

with open('filelist.txt') as f:
    paths = [line.strip() for line in f if line.strip()]

with zipfile.ZipFile('project2.zip', 'w', zipfile.ZIP_DEFLATED) as zf:
    for path in paths:
        zf.write(path)

print('done')
!ls -la project2.zip
```
